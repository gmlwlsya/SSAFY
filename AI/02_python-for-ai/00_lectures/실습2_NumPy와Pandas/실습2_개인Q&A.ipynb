## 1. NumPy와 Pandas는 언제 구분해서 쓸까?

두 라이브러리 모두 데이터를 다루지만, **데이터의 형태와 목적**에 따라 쓰임새가 다릅니다.

| 구분 | **NumPy (넘파이)** | **Pandas (판다스)** |
| --- | --- | --- |
| **주요 대상** | **숫자 중심의 다차원 배열 (`ndarray`)** | **표 형태의 데이터 (`DataFrame`, `Series`)** |
| **데이터 타입** | **단일 타입** (모든 요소가 숫자 or 문자) | **혼합 타입** (숫자, 문자, 날짜 등 열마다 다름) |
| **주요 용도** | 복잡한 수학/수치 계산, 행렬 연산, 머신러닝 기초 연산 | 엑셀 데이터 분석, CSV 파일 읽기/쓰기, 데이터 정제 및 그룹화 |
| **비유하자면** | 공학용 계산기 & 수학 연산 전용 도구 | 파이썬판 고급 엑셀(Excel) |

### 💡 실무에서는 어떻게 쓰나요?

* **Pandas를 쓸 때:** 엑셀이나 DB처럼 "이름, 나이, 이메일, 결제금액"같이 **다양한 타입의 데이터가 섞여 있는 표**를 다룰 때 씁니다.
* **NumPy를 쓸 때:** 순수하게 **숫자로만 이루어진 대용량 데이터 계산**이나 복잡한 선형대수(행렬) 연산을 엄청 빠른 속도로 처리해야 할 때 씁니다.

## 2. NumPy는 `np.sort(array)`인데 Pandas는 `array.sort()`처럼 쓰면 안 되나요?

결론부터 말씀드리면, **NumPy 배열(array)에는 `.sort()`라는 메서드가 이미 존재합니다!** 하지만 동작 방식(원본이 바뀌는가?)에 결정적인 차이가 있습니다.

또한, Pandas의 데이터(Series, DataFrame)에서는 `.sort()`가 아니라 `.sort_values()`라는 이름을 사용합니다.

### 1) NumPy 배열에서의 정렬 차이

NumPy에서는 두 가지 방식이 모두 가능하지만 동작이 다릅니다.

* **`np.sort(arr)` (함수형):**
* **원본을 유지**하고, 정렬된 **새로운 배열을 반환**합니다.


```python
import numpy as np
arr = np.array([3, 1, 2])
sorted_arr = np.sort(arr)

print(sorted_arr) # [1, 2, 3] (새로운 배열)
print(arr)        # [3, 1, 2] (원본은 그대로!)

```


* **`arr.sort()` (메서드형):**
* **원본 자체를 직접 정렬**해버립니다. (반환값은 `None`)


```python
arr = np.array([3, 1, 2])
arr.sort()

print(arr)        # [1, 2, 3] (원본이 변경됨!)

```



---

### 2) Pandas 데이터에서의 정렬 방식

Pandas의 `Series`나 `DataFrame` 객체는 `sort()` 대신 더 명확한 이름인 `sort_values()`나 `sort_index()`를 사용합니다.

```python
import pandas as pd

# Pandas Series 예시
s = pd.Series([3, 1, 2])

# 값 기준으로 정렬 (새로운 Series 반환)
sorted_s = s.sort_values() 

print(sorted_s)
# 1    1
# 2    2
# 0    3

```

> **💡 요약하자면:**
> 1. NumPy 배열에도 `arr.sort()`를 쓸 수 있지만, **원본 데이터가 바로 바뀌어 버리므로** 원본을 보존하고 싶을 땐 `np.sort(arr)`를 씁니다.
> 2. Pandas 객체에서는 `.sort()` 대신 **`.sort_values()`**를 사용합니다.

# [:, np.newaxis]
NumPy에서 `[:, np.newaxis]`는 "기존 데이터의 차원을 한 단계 늘려서 2차원의 열(Column) 벡터 형태로 변환하겠다"는 의미를 가진 문법입니다.

처음 보면 생소하고 가독성이 떨어져 보일 수 있지만, 백엔드 데이터 처리나 머신러닝 연산에서 매우 자주 쓰이는 핵심 스킬 중 하나입니다.

---

## 🔍 하나씩 쪼개서 이해하기

`[:, np.newaxis]`는 슬라이싱 기호인 **`:`** 와 **`np.newaxis`** 의 합성입니다.

1. **`:` (콜론):** "기존 차원의 모든 요소(모든 행)를 그대로 가져오겠다"는 뜻입니다.
2. **`np.newaxis`:** "이 자리에 새로운 차원(축, Axis)을 하나 추가하겠다"는 뜻입니다.
* `None`과 완전히 동일하게 작동합니다. (즉, `arr[:, None]`으로 써도 같습니다.)



따라서 `arr[:, np.newaxis]`는 "1번째 축(행)은 기존 데이터를 다 유지하고, 2번째 축(열)을 새롭게 만들어서 넣어라"라는 지시가 됩니다.

---

## 💻 코드 예시로 비교하기

### 1. 1차원 배열 준비

```python
import numpy as np

a = np.array([10, 20, 30])

print(a)
# 출력: [10 20 30]

print(a.shape)
# 출력: (3,)  <-- 1차원 (원소 3개짜리 벡터)

```

### 2. `[:, np.newaxis]` 적용 후

```python
a_2d = a[:, np.newaxis]

print(a_2d)
# 출력:
# [[10]
#  [20]
#  [30]]

print(a_2d.shape)
# 출력: (3, 1)  <-- 2차원 (3행 1열짜리 행렬)

```

## 💡 어디에 쓰이고 왜 중요한가요? (사용 이유)

### 1. 머신러닝/딥러닝 모델의 입력값 맞추기

Scikit-learn이나 PyTorch, TensorFlow 같은 머신러닝 라이브러리의 모델들은 입력 데이터(Feature)로 **2차원 배열 `(N개의 샘플 수, Feature 수)**` 형태를 요구합니다.

* 특성(Feature)이 1개일 때, `[1, 2, 3]`(1차원)을 그대로 넣으면 에러가 납니다.
* 이때 `[:, np.newaxis]`를 써서 `[[1], [2], [3]]`(2차원, (3, 1) 형태)으로 만들어 전달할 때 아주 유용합니다.

### 2. 브로드캐스팅(Broadcasting) 연산

행렬 간의 사칙연산을 할 때 차원(Shape)이 맞지 않으면 계산이 불가능합니다.
차원을 맞춰주어 1차원 배열과 2차원 배열 간의 자동 확장 연산(브로드캐스팅)이 가능하게 만들어 줍니다.

## 📌 똑같은 기능을 하는 다른 표현들

취향이나 상황에 따라 아래 표현들을 혼용해서 사용하지만, 모두 동일한 결과를 만듭니다.

```python
# 1. np.newaxis 사용 (가장 명시적)
a_2d = a[:, np.newaxis]

# 2. None 사용 (np.newaxis 대신 None을 적음, 짧아서 자주 씀)
a_2d = a[:, None]

# 3. reshape() 사용 (가장 직관적인 함수형)
a_2d = a.reshape(-1, 1)

```

> **💡 요약:** `[:, np.newaxis]`는 **`(N,)` 모양의 1차원 배열을 `(N, 1)` 모양의 2차원 열 벡터로 변환하는 가장 빠른 테크닉**입니다!
